## 1. Two processes

When you run a notebook, two programs work together:

```
 Python kernel                               Browser (JupyterLab, VS Code, Colab)
 ┌──────────────────────────┐   messages    ┌──────────────────────────────┐
 │ widget object            │ ◄───────────► │ widget model ──► view (cell A) │
 │   value = 5   (traits)   │   ("comm")    │   value = 5  ──► view (cell B) │
 └──────────────────────────┘               └──────────────────────────────┘
```

A **widget** is a Python object whose traits (attributes) are copied to a
**model** in the browser. When either side changes a synced trait, the other
side gets a message and updates.

````python
import ipywidgets as widgets
from IPython.display import display

slider = widgets.IntSlider(value=5, min=0, max=10, description="value")
slider
````

## 3. Under the hood

`slider.keys` lists the synced traits. `get_state()` is exactly what travels
between the kernel and the browser.

````python
print("model id:", slider.model_id)
print("synced traits:", len(slider.keys))
state = slider.get_state()
{key: state[key] for key in ["value", "min", "max", "description", "_model_name", "_view_name"]}
````

## 4. React to changes with `observe`

`observe` calls your function every time a trait changes, no matter which side
changed it.

> **Important:** `print()` inside a callback does not appear in the cell in
> JupyterLab; it goes to the log console. Write to an `Output` widget instead.

````python
log = widgets.Output()


def on_value(change):
    with log:
        log.clear_output(wait=True)
        print(f"value changed from {change['old']} to {change['new']}")


slider.observe(on_value, names="value")
display(slider, log)
````

## 5. Connect widgets with `link`

- `widgets.link` syncs two traits **through the kernel**.
- `widgets.jslink` syncs them **in the browser only**. It also works in a
  saved HTML page with no kernel.

````python
amplitude = widgets.FloatSlider(value=1.0, min=0, max=5, description="slider")
amplitude_box = widgets.FloatText(description="text box")
widgets.link((amplitude, "value"), (amplitude_box, "value"))
widgets.VBox([amplitude, amplitude_box])
````

| 2015 | "The Big Split": widgets move to their own package, **ipywidgets** |
| 2015 to 2022 | Custom widgets (bqplot, ipyleaflet, ipyvolume) need Python + npm + webpack builds |
| 2023 | **anywidget**: a custom widget is one Python class plus one JavaScript file |
| 2024 | The Jupyter widget template is deprecated in favour of anywidget |
| 2026 | anywidget 0.11 adds widget composition |

# Station A: jupyter-scatter

**Good for:** scatter plots with up to millions of points (embeddings,
single-cell data), with pan, zoom, lasso, and smooth transitions. Uses WebGL.
Works offline.

Docs: https://jupyter-scatter.dev

```python
import ipywidgets as widgets
import jscatter
import numpy as np
import pandas as pd
from sklearn.datasets import make_blobs

X, labels = make_blobs(n_samples=200_000, centers=12, n_features=2, cluster_std=1.2, random_state=3)
df = pd.DataFrame({
    "x": X[:, 0],
    "y": X[:, 1],
    "cluster": pd.Categorical(labels.astype(str)),
    "score": np.random.default_rng(0).normal(size=len(X)),
})
f"{len(df):,} points"
```

**Lasso:** hold **Shift** and drag. Scroll to zoom. Double-click to reset.

```python
scatter = jscatter.Scatter(data=df, x="x", y="y", color_by="cluster", height=450)
scatter.show()
```

Get the selection back into Python:

```python
selected = df.iloc[scatter.selection()]
selected.groupby("cluster", observed=True).size().sort_values(ascending=False).head()
```

Select from Python, for example the 1,000 points with the highest score:

```python
scatter.selection(df["score"].nlargest(1000).index.to_numpy())
```

### Your task: connect it

1. A dropdown that switches colouring between `cluster` and `score`.
   (`scatter.color(by="score", map="viridis")`)
2. An `Output` that shows the mean `score` of the selection every time it changes.
   (`scatter.widget.observe(callback, names=["selection"])`)

```python
color_choice = widgets.Dropdown(options=["cluster", "score"], description="colour by")
summary = widgets.Output()


def on_color(change):
    if change["new"] == "score":
        scatter.color(by="score", map="viridis")
    else:
        scatter.color(by="cluster")


def on_selection(change):
    rows = df.iloc[scatter.selection()]
    with summary:
        summary.clear_output(wait=True)
        print(f"{len(rows):,} selected, mean score {rows['score'].mean():.3f}")


color_choice.observe(on_color, names="value")
scatter.widget.observe(on_selection, names=["selection"])
widgets.VBox([color_choice, summary])
```

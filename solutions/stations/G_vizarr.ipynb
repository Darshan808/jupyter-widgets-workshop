# Station G: vizarr

**Good for:** viewing very large images (microscopy, OME-Zarr) by loading only
the tiles you look at. From the HIDIVE lab at Harvard Medical School.
**Needs internet:** the viewer is loaded in the browser.

Docs: https://github.com/hms-dbmi/vizarr

```python
import ipywidgets as widgets
import numpy as np
import zarr
import vizarr

# A synthetic "fluorescence" image: 2,048 x 2,048 pixels with 400 bright cells.
rng = np.random.default_rng(3)
size = 2048
yy, xx = np.mgrid[0:size, 0:size]
image = rng.normal(100, 10, (size, size))
for cy, cx in rng.integers(0, size, (400, 2)):
    r = rng.integers(8, 20)
    window = (slice(max(cy - 3 * r, 0), cy + 3 * r), slice(max(cx - 3 * r, 0), cx + 3 * r))
    image[window] += 2000 * np.exp(-((yy[window] - cy) ** 2 + (xx[window] - cx) ** 2) / (2 * r**2))
image = image.clip(0, 65535).astype(np.uint16)

array = zarr.array(image, chunks=(256, 256))
f"{array.shape}, {array.nchunks} chunks"
```

```python
viewer = vizarr.Viewer()
viewer.add_image(source=array, name="cells")
viewer
```

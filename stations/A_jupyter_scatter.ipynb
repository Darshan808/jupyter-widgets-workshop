# Station A: jupyter-scatter

**Good for:** scatter plots with up to millions of points (embeddings,
single-cell data), with pan, zoom, lasso, and smooth transitions. Uses WebGL.
Works offline.

Docs: https://jupyter-scatter.dev

```python
import ipywidgets as widgets
import jscatter
import numpy as np
import pandas as pd
from sklearn.datasets import make_blobs

X, labels = make_blobs(n_samples=200_000, centers=12, n_features=2, cluster_std=1.2, random_state=3)
df = pd.DataFrame({
    "x": X[:, 0],
    "y": X[:, 1],
    "cluster": pd.Categorical(labels.astype(str)),
    "score": np.random.default_rng(0).normal(size=len(X)),
})
f"{len(df):,} points"
```

**Lasso:** hold **Shift** and drag. Scroll to zoom. Double-click to reset.

```python
scatter = jscatter.Scatter(data=df, x="x", y="y", color_by="cluster", height=450)
scatter.show()
```

Get the selection back into Python:

```python
selected = df.iloc[scatter.selection()]
selected.groupby("cluster", observed=True).size().sort_values(ascending=False).head()
```

Select from Python, for example the 1,000 points with the highest score:

```python
scatter.selection(df["score"].nlargest(1000).index.to_numpy())
```

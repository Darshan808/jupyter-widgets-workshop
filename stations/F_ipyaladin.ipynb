# Station F: ipyaladin

**Good for:** exploring sky surveys and catalogues, from the Strasbourg
astronomical data centre (CDS). **Needs internet** for the sky images.
Use coordinates (not names) as targets: name lookup also needs the internet.

Docs: https://cds-astro.github.io/ipyaladin

```python
import ipywidgets as widgets
from ipyaladin import Aladin

aladin = Aladin(target="10.6847 41.2690", fov=3)  # the Andromeda galaxy (M31)
aladin
```

### Your task: connect it

Make a dropdown of famous objects (coordinates below) that moves the view, and
a slider that sets the field of view in degrees.

```python
targets = {
    "Andromeda galaxy (M31)": "10.6847 41.2690",
    "Orion nebula (M42)": "83.8221 -5.3911",
    "Crab nebula (M1)": "83.6331 22.0145",
    "Galactic centre": "266.4168 -29.0078",
}
choice = widgets.Dropdown(options=list(targets), description="target")
fov = widgets.FloatSlider(value=3, min=0.1, max=30, step=0.1, description="fov (deg)", continuous_update=False)

# TODO: observe choice -> set aladin.target; observe fov -> set aladin.fov

widgets.VBox([choice, fov])
```
